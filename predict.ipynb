```python
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import tensorflow as tf
```

```python
with open('label_encoder.pkl', 'rb') as f:
    label_encoder = pickle.load(f)
    
```

```python
with open('onehot_encoder.pkl', 'rb') as f:
    onehot_encoder = pickle.load(f)
```

```python
with open('standard_scaler.pkl', 'rb') as f:
    standard_scaler = pickle.load(f)
```

```python
loaded_model = load_model('model.h5')
```

```python
user = {'Pclass': 3, 'Sex': 'female', 'Age': 25, 'SibSp': 1, 'Parch': 1, 'Fare': 7.25, 'Embarked': 'S'}
```

```python
user_df = pd.DataFrame([user])
```

```python
user_df
```

```python
user_df['Sex'] = label_encoder.transform(user_df[['Sex']])
```

```python
user_df
```

```python
embarked_encoded = onehot_encoder.transform(user_df[['Embarked']])
```

```python
pd.DataFrame(embarked_encoded, columns=onehot_encoder.get_feature_names_out(['Embarked']))
```

```python
user = pd.concat([user_df.drop('Embarked', axis=1), pd.DataFrame(embarked_encoded, columns=onehot_encoder.get_feature_names_out(['Embarked']))], axis=1)
```

```python
user
```

```python
# Check which features the scaler expects
print(standard_scaler.feature_names_in_)

# Pass all expected columns to transform
scaler_features = standard_scaler.feature_names_in_
user[scaler_features] = standard_scaler.transform(user[scaler_features])
```

```python
user
```

```python
y = loaded_model.predict(user)
```

```python
y =y[0][0]
```

```python
y
```

```python
if y>= 0.5:
    print("The passenger is predicted to survive.")
else:
    print("The passenger is predicted not to survive.")
```
